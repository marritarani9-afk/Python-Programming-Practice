```python
# 29. Check whether a number is a palindrome

n = int(input("Enter a number: "))

if n < 0:
    print("Not a palindrome")
else:
    original = n
    reverse = 0

    while n > 0:
        digit = n % 10
        reverse = reverse * 10 + digit
        n //= 10

    if original == reverse:
        print("Palindrome")
    else:
        print("Not a palindrome")
```

```python
# 34. Enter a number and print it in words

n = int(input("Enter a number (0-9999): "))

ones = ["", "One", "Two", "Three", "Four", "Five",
        "Six", "Seven", "Eight", "Nine"]

teens = ["Ten", "Eleven", "Twelve", "Thirteen", "Fourteen",
         "Fifteen", "Sixteen", "Seventeen", "Eighteen", "Nineteen"]

tens = ["", "", "Twenty", "Thirty", "Forty", "Fifty",
        "Sixty", "Seventy", "Eighty", "Ninety"]


def below_thousand(num):
    words = []

    if num >= 100:
        words.append(ones[num // 100] + " Hundred")
        num %= 100

    if num >= 20:
        words.append(tens[num // 10])
        num %= 10
    elif num >= 10:
        words.append(teens[num - 10])
        num = 0

    if num > 0:
        words.append(ones[num])

    return " ".join(words)


if n < 0 or n > 9999:
    print("Number out of range")
elif n == 0:
    print("Zero")
else:
    words = []

    if n >= 1000:
        words.append(ones[n // 1000] + " Thousand")
        n %= 1000

    if n > 0:
        words.append(below_thousand(n))

    print(" ".join(words))
```

```python

# 43. Check whether a number is an Armstrong number

n = int(input("Enter a number: "))

if n < 0:
    print("Not an Armstrong number")
else:
    digits = str(n)
    power = len(digits)
    total = 0

    for digit in digits:
        total += int(digit) ** power

    if total == n:
        print("Armstrong number")
    else:
        print("Not an Armstrong number")
```

```python
# 83. Check whether a number is Prime, Armstrong or Perfect using functions

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


def is_armstrong(n):
    digits = str(n)
    total = 0

    for digit in digits:
        total += int(digit) ** len(digits)

    return total == n


def is_perfect(n):
    total = 0

    for i in range(1, n):
        if n % i == 0:
            total += i

    return n > 0 and total == n


num = int(input("Enter a number: "))

print("Prime:", is_prime(num))
print("Armstrong:", is_armstrong(num))
print("Perfect:", is_perfect(num))
```

```python
# 89. Display ATM transactions

balance = 10000

while True:
    print("\n1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        print("Balance:", balance)

    elif choice == 2:
        amount = int(input("Enter deposit amount: "))

        if amount > 0:
            balance += amount
            print("Amount deposited successfully")
            print("Balance:", balance)
        else:
            print("Invalid amount")

    elif choice == 3:
        amount = int(input("Enter withdrawal amount: "))

        if amount <= 0:
            print("Invalid amount")
        elif amount <= balance:
            balance -= amount
            print("Please collect your cash")
            print("Balance:", balance)
        else:
            print("Insufficient balance")

    elif choice == 4:
        print("Thank you for using ATM")
        break

    else:
        print("Invalid choice")

```
